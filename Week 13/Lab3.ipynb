## Data Preprocessing

**Step 1 - Import Libraries**

```python
%pip install tensorflow
import numpy as np
import pandas as pd
import tensorflow as tf
```

**Step 2 - Load Data**

```python
dataset = pd.read_csv('Churn_Modelling.csv')
X = dataset.iloc[:, 3:-1].values
y = dataset.iloc[:, -1].values
```

Inspect data (X)

```python
print(X)
```

**Step 3 - Encode Categorical Data**

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:, 2] = le.fit_transform(X[:, 2])
```

Check data (X) with print. The result,

```python
print(X)
```

**Step 4 - Encode the "Geography" Column with One Hot Encoder**

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

Check data (X) with print. The result,

```python
print(X)
```

**Step 5 - Split Data**

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

**Step 6 - Feature Scaling**

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Constructing the ANN Model

**Step 1 - Initialize the ANN Model**

```python
ann = tf.keras.models.Sequential()
```

**Step 2 - Create the Input Layer and the First Hidden Layer**

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

**Step 3 - Create the Second Hidden Layer**

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

**Step 4 - Create the Output Layer**

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

## Training the Model

**Step 1 - Compile the ANN Model (Unify the Architecture)**

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

**Step 2 - Fit the Model**

```python
ann.fit(X_train, y_train, batch_size = 32, epochs = 100)
```

## Making Predictions

Given the following information,

- Geography: France

- Credit Score: 600

- Gender: Male

- Age: 40 years old

- Tenure: 3 years

- Balance: $ 60000

- Number of Products: 2

- Does this customer have a credit card? Yes

- Is this customer an Active Member: Yes

- Estimated Salary: $ 50000

Should this customer be retained?

**Model the New Data and Make a Prediction**

```python
print(ann.predict(sc.transform([[1, 0, 0, 600, 1, 40, 3, 60000, 2, 1, 1, 50000]])) > 0.5)
```

Is the result **False**?

Yes, the result is False. That means the ANN predicts this customer will not leave. So the customer should be retained (because they are not at risk).

**Predictions with the Testing Data**

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

**Evaluate Accuracy and the Confusion Matrix**

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
